# Variable, Data & Data types


A variable is an assigned memory where any type of value is stored and each value stored in the variable is a data. One or more single data combine to form a data group, similarly a dataset is formed with one or more data groups. In general, we know data, data group and dataset all of them as data. The characteristic classification of the data stored in the variable is called data type. Here we will discuss about character or string and numeric (integer & float) data types.


### A variable is a labelled box

```
x   =   5
↑   ↑   ↑
name  assignment  value (data)
      operator
```

The `=` sign is **not** the "equals" of mathematics — it means **"store the value on the right under the name on the left"**.
To compare two things you use `==`, which comes in the next class.

### Assign and re-assign

If you give the same name a new value, the old one is gone. Python always remembers only the most recent value.

````python
x = 5
print(x)
````

````python
x = 10        # same name, new value
print(x)
````

````python
# Every cell in Jupyter shares the same memory —
# so x is available from a different cell too.

print(x)
````

### Naming variables: rules and conventions

**Rules — enforced by Python:**
- Only letters, digits and the underscore `_` are allowed
- A name can never start with a digit
- No spaces inside a name
- Python is case-sensitive: `Age`, `age`, `AGE` are three different variables
- Reserved keywords cannot be used as names: `if`, `for`, `class`, `True`, `import` …

**Conventions — PEP 8, Python's official style guide:**
- Use `snake_case`: `int_number`, `total_marks`, `str_1`
- Choose meaningful names — reading the name should tell you what is inside
- Start with a lowercase letter
- Avoid a lone `l`, `O` or `I` — they look like `1` and `0`

````python
# Valid names
int_number  = 24
total_marks = 87.5
str_1       = 'Dhaka'

print(int_number, total_marks, str_1)
````

### Type conversion (casting) — very important

Notice: above, `number = '250.40'` is written inside quotation marks, so it is a **str**, not a number.
It looks like a number, but Python cannot do arithmetic with it.

| Function | What it does |
|----------|-----|
| `int()` | text or float → whole number (it truncates the decimal part, it does not round) |
| `float()` | text or int → decimal number |
| `str()` | any value → text |

> The most common mistake in data analysis — numbers that are actually stored as text.
> This happens very often when reading data from CSV or Excel.

````python
number = '250.40'        # this is a string

print(number + '10')     # the text was joined, nothing was added
print(type(number))
````

````python
value = float(number)    # string -> float conversion

print(value + 10)        # now it is real addition
print(type(value))
````

````python
print(int('24'),   type(int('24')))
print(str(24),     type(str(24)))
print(int(5.9))                      # 5 — truncates the decimal, does not round

# print(int('5.7'))                  # ValueError! uncomment to see it
print(int(float('5.7')))             # the correct way: float first, then int
````

### Presenting results nicely: the f-string

Put an `f` before the opening quote and you can write any variable inside `{ }`.
`:.2f` controls how many decimal places are shown.

````python
name  = 'Wabidur'
marks = 87.5

print('Name :', name, 'Marks :', marks)       # with commas — Python inserts the spaces itself
print(f'Name : {name}, Marks : {marks}')      # f-string
print(f'Marks rounded : {marks:.1f}')         # one decimal place
````

| Error | Cause | Fix |
|-------|------|--------|

### Arithmatic operators
- Addition (+) $\rightarrow$ x+y
- Subtraction (-) $\rightarrow$ x-y
- Multiplication () $\rightarrow$ x*y
- Division (/) $\rightarrow$ x/y (It is also known as true division)
- Floor division (//) $\rightarrow$ x//y
- Modulus (%) $\rightarrow$ x%y (It is also known as reminder)
- Exponentiation () $\rightarrow$ x**y (It is also known as power)


````python
a = 50
b = 5.7

add = a+b
sub = a-b
mul = a*b
true_div = a/b


print('Addition : ', add)
print('Subtraction : ', sub)
print('Multiplication : ', mul)
print('True division : ', true_div)
````

````python
floor_div = a//b
modulus = 5%2
power = 10**3

print('Floor division : ', floor_div)
print('Reminder or modulus : ', modulus)
print('Exponent or power : ', power)
````

### Why the long tail of decimals?

Computers store decimal numbers in binary, so a tiny rounding error remains in some calculations.
This is not a Python bug — every programming language behaves this way.
Use `round()` when you need a tidy result.

````python
print(0.1 + 0.2)              # 0.30000000000000004

a = 50
b = 5.7
print(a / b)                  # 8.771929824561402
print(round(a / b, 2))        # 8.77
````

### Operator precedence — which runs first?

| Order | Operator | Name |
|------|----------|-----|
| 1 | `( )` | Parentheses — always first |
| 2 | `**` | Exponent (right to left) |
| 3 | `-x` | Negative sign |
| 4 | `*` `/` `//` `%` | Multiply and divide (left to right) |
| 5 | `+` `-` | Add and subtract (left to right) |

This is the familiar BODMAS rule of mathematics.
**A safe habit:** when there are more than two operations, write the parentheses yourself.

````python
print(10 - 2 * 3)       # 4  — multiplication happens first
print((10 - 2) * 3)     # 24 — the parentheses happen first
````

**Symbols and their matching functions:**

| Symbol | Function | Symbol | Function |

**Why learn both?** For everyday arithmetic the symbols are simpler.
But when an operation itself has to be passed as an argument (to sorting, `map()`, `reduce()`), the function form is what you need.

#### Root
problem: $\rightarrow$ $\sqrt{x+x^2}-1$


````python
# Root or sqrt methods

import math as mt

x = 25
result_1 = mt.sqrt(x)
result_2 = mt.sqrt(x+x**2) -1

print('Result 1 : ', result_1)
print('Result 2 : ', result_2)
````

````python
add_1 = op.add(x, op.pow(x, 2))
sq = mt.sqrt(add_1)
sub_1 = op.sub(sq, 1)
print(sub_1)
````

`mt.sqrt(25)` and `25 ** 0.5` both give `5.0`.
So the exponent operator can take a square root too, without importing any module.

A few more useful functions from the `math` module: `mt.pi`, `mt.floor()`, `mt.ceil()`.

````python
import math as mt

print(mt.sqrt(25), 25 ** 0.5)
print(mt.pi)
print(mt.floor(5.7), mt.ceil(5.2))
````

### Assignment operator
Assignment operator is the short form in written of others.


| Operator | Example | Equivalent to |
|----------|--------|---------|
| `=`  | `x = 5`   | store 5 in x |
| `+=` | `x += 3`  | `x = x + 3` |
| `-=` | `x -= 3`  | `x = x - 3` |
| `*=` | `x *= 3`  | `x = x * 3` |
| `/=` | `x /= 3`  | `x = x / 3` |
| `//=`| `x //= 3` | `x = x // 3` |
| `%=` | `x %= 3`  | `x = x % 3` |
| `**=`| `x **= 3` | `x = x ** 3` |

An assignment operator is not a new calculation — it is a shorthand for a calculation you already know.

````python
a = 50
b = 5.7

add = a+b
sub = a-b
mul = a*b
true_div = a/b

print(add)
````

````python
add = add + 10
print(add)
````

````python
add += 10
print(add)
````

````python
x = 2

add += 2
sub -= x
mul *= x
true_div /= x

print('Addition : ', add)
print('Subtraction : ', sub)
print('Multiplication : ', mul)
print('True division : ', true_div)
````

````python
floor_div //= x
modulus %= x
power **= x

print('Floor division : ', floor_div)
print('Reminder or modulus : ', modulus)
print('Exponent or power : ', power)
````
