# Part 1 — Function

### Where is the problem?

If the same calculation has to be done again and again, the same code has to be written again and again.
A **function** ties that code to a name — write it once, use it as often as you like.

```
def function_name(parameters):
    'function description'
    statement(s)
    return value        <- optional

function_name(arguments)   <- calling the function
```

Remember two words —
- **`def`** = define, the word that creates a function
- **`return`** = sending a result back out of the function (optional)

````python
# the simplest function — it does a job and returns nothing

def text():
    print('I am a Python Programmer.')

text()          # without the call, the code inside never runs
````

### Sending a result back with `return`

`print` only shows something on screen. But to use the result later you must send it back
with `return`. Then the value can be stored in a variable.

````python
# returns the mean of three numbers

def avr(x, y, z):
    add = x + y + z
    result = add / 3
    return result

answer = avr(3, 4, 5)      # storing the returned value
print('mean =', answer)
````

### Parameter and Argument

- **Parameter** — the name inside the brackets when the function is written (`x, y, z`)
- **Argument** — the actual value passed when it is called (`3, 4, 5`)

Put simply — a parameter is the empty box, an argument is what you put in it.

````python
# takes one argument, returns an f-string

def make_name(n):
    return f'My name is {n}'

print(make_name('Rahman'))
print(make_name('Wabidur'))
````

### Returning several values at once

Put a few values after `return`, separated by commas, and Python packs them into a **tuple** for you.

````python
# returning several results at once

def calculate(a, b, c):
    add = a + b
    sub = a - b
    mul = a * c
    result = add + mul - sub
    return result, add, mul, sub      # four values at once

print(calculate(3, 4, 5))
````

### Default arguments — a value decided in advance

Give a parameter a value with `=` and you no longer have to pass it when calling —
the default is used instead.

````python
# y and z have default values

def avr2(x, y=2, z=3):
    add = x + y + z
    return add, add / 3

print('all three given :', avr2(3, 4, 5))  # y=4, z=5 are used
print('only x given    :', avr2(3))        # the defaults y=2, z=3 are used
````

### `*args` — an unknown number of arguments

When you do not know how many arguments will arrive, put a `*` before the parameter. Every
argument then arrives as a **tuple**, and you can reach them by index.

````python
# pass as many names as you like — they all arrive in one tuple

def my_fun(*kids):
    print('all names :', kids)
    print('2nd one   :', kids[1])      # index 1

my_fun('Abid', 'Raiyan', 'Fatima', 'Wabidur')
````

### `print()` has parameters of its own

`print` is itself a function. Its `sep` (what goes between items) and `end` (what goes at the end)
parameters can be changed. Until now we have been using the defaults — `sep=' '`, `end='\n'`.

````python
# changing sep and end

print('I am a programmer', 'My name is Wabidur', sep=' ___ ', end=' *** ')
print('This is Python course')     # this sits right after the *** of the previous line
print('ABC')
````

### Function + loop — a machine for computing means

The accumulator technique from Class 04, now inside a function. Build it once and you can take the
mean of any list — even of a pandas column.

````python
# returns the mean and the sum of a list

def mean_f(li):
    """li is a list of numbers"""
    add = 0
    for i in li:
        add = add + i
    result = f'mean {add / len(li)}'
    total = f'sum {add}'
    return result, total


list_1 = [1, 2, 3, 4, 5, 2, 4, 5, 3, 4, 56, 4, 2, 5]
print(mean_f(list_1))
````

````python
# the same function also works on a pandas column

import pandas as pd

df = pd.read_csv('sample1.csv')
df.columns = [c.replace('\n', ' ').strip() for c in df.columns]   # clean the names

p = df.Probability
print(mean_f(p))

# checking against a ready-made module
import numpy as np
print('numpy sum :', np.sum(p))
````

### Recursion — a function that calls itself

A function can call **itself from inside itself** — this is called recursion.
The best known example is the **factorial**: 5! = 5 × 4 × 3 × 2 × 1.

Two things are essential — a **stopping condition** (the base case), otherwise it runs forever.

````python
# factorial — with recursion

def factorial(n):
    if n <= 1:              # the stopping condition (base case)
        return 1
    return n * factorial(n - 1)     # calling itself

for n in [1, 3, 5, 6]:
    print(f'{n}! =', factorial(n))
````

# Part 2 — Lambda Function

A **lambda** is an anonymous function — it fits on a single line.

```
lambda arguments : expression
```

- It can take any number of arguments, but it can hold **only one expression**
- You do not write `return` — the value of the expression is returned automatically
- When you need a small function for a moment, a lambda is the most convenient option

````python
# the same job — first with def, then with lambda

def double(a):          # an ordinary function
    return a * 2

l_d = lambda a: a * 2   # the same job, in one line

print('with def    :', double(5))
print('with lambda :', l_d(5))
````

````python
# it can take several arguments too

y = lambda a, b: a + b
multiple = lambda a, b, c: a + b + c

print(y(5, 10))
print(multiple(3, 4, 5))
````

### Where a lambda really shines — inside another function

A lambda on its own is not much of a gain. Its real use is inside functions such as `sorted()`,
`map()` and `filter()` — where you need to state a small rule on the spot.

````python
# sorted() — ranking students by mark

students = [('Rahim', 82), ('Karim', 55), ('Sadia', 90)]

# key=lambda says: sort by the 2nd item of the pair (the mark), not the name
ranked = sorted(students, key=lambda s: s[1], reverse=True)

print('by mark, highest to lowest:')
for name, mark in ranked:
    print(f'  {name:<8} {mark}')
````

````python
# map() — the same job on every item

squares = list(map(lambda x: x ** 2, [1, 2, 3, 4]))
print('squares :', squares)

# filter() — picking the items that match a condition
evens = list(filter(lambda x: x % 2 == 0, range(10)))
print('evens   :', evens)
````

> **When lambda, when def?** The job is small, one line, and needed once → lambda.
> The job is big, several lines, or needed repeatedly → a named function with `def`.

# Part 3 — String Methods

Working with text is a large part of data analysis — cleaning names, searching, replacing,
splitting. In Python a string is created with **single** `' '` or **double** `" "` quotes.

````python
# creating a string and checking its type

text = 'This is a Python class.'
print(text)
print('type :', type(text))
````

### A common mistake — a quote inside a quote

Write `'I'm a student'` and Python reads the string from the first `'` to the second `'` (right after I),
then cannot make sense of the rest and raises a **SyntaxError**. The mistake is shown safely below.

````python
# running the mistake on purpose to see the message

broken = "text = 'I'm a student'"

try:
    exec(broken)
except SyntaxError as e:
    print('❌ SyntaxError:', e.msg)
````

### The fix — escape (`\`) or the other quote

Three ways —
1. If there is a `'` inside, use `"` outside (or the other way round)
2. Escape it by writing `\'`
3. `\n` = newline, `\t` = tab — these are escape characters too

````python
# three ways

text_1 = "I'm a Researcher."                          # double quotes outside
text_2 = 'She said "Hello".'                           # single quotes outside
text_3 = 'Line one.\nLine two.'                        # \n = newline

print(text_1)
print(text_2)
print(text_3)
````

### Concatenation and slicing (just as in Class 03)

`+` joins strings, and `[start:end]` takes a piece — exactly like a List.

````python
# joining

a, b, c = 'Md', 'Wabidur', 'Rahman'

print(a + b + c)             # without a space
print(a + ' ' + b + ' ' + c) # with a space
print(a, b, c)               # print adds the space itself
````

````python
# join() — a string from a list

li = ['Md', 'Wabidur', 'Rahman']
print(' '.join(li))          # joined with a space
print('*'.join(li))          # joined with a star

# slicing
text = 'Researcher'
print(text[0:6])             # the first six characters
print(text[::-1])            # reversed
````

### Searching and replacing — `find()`, `replace()`, `count()`

````python
# find() — where it is (index), replace() — swapping it out

text_1 = 'I am a Researcher. My name is Md Wabidur Rahman.'

print('where is name :', text_1.find('name'))
print('replaced      :', text_1.replace('Rahman', 'Khan'))

# count() — how many times it occurs
s5 = 'this is a test. this is another test. this is final test.'
print('test occurs   :', s5.count('test'), 'times')
````

### Removing whitespace — the `strip()` family

Real data often has unwanted spaces before and after a name. `strip()` removes them from both sides,
`lstrip()` from the left, `rstrip()` from the right.

````python
# strip() — three ways to remove whitespace

text = ' This is a string. '

print(f'original: [{text}] length {len(text)}')
print(f'strip   : [{text.strip()}] length {len(text.strip())}')
print(f'lstrip  : [{text.lstrip()}] length {len(text.lstrip())}')
print(f'rstrip  : [{text.rstrip()}] length {len(text.rstrip())}')
````

### Upper and lower case — changing case

In Class 04 we matched country names with `.title()`, remember? Here is the whole family —

````python
# four ways to change case

s2 = 'i am a researcher. my name is md wabidur rahman.'

print('upper      :', s2.upper())
print('lower      :', s2.lower())
print('capitalize :', s2.capitalize())     # only the first letter capitalised
print('title      :', s2.title())          # the first letter of every word capitalised
````

### Splitting with `split()`, and checking with `startswith()` / `endswith()`

````python
# split() — breaks on a character and builds a list

str_5 = 'Md_Wabidur_Rahman'
print("split on '_' :", str_5.split('_'))

# startswith / endswith — return True/False
s5 = 'this is a test.'
print('starts with this? :', s5.startswith('this'))
print('ends with test.? :', s5.endswith('test.'))
````

### A practical use — addressing someone by their name

`startswith()` plus `if–elif–else` lets you address a user correctly.
It is shown below with fixed names, and the `input()` version is left commented out.

> **⚠ The Class 04 lesson applies here too:** `'mrs'` must be checked **before** `'mr'` —
> because `'mrs sadia'` also starts with `'mr'`! Reverse the order and you will call a Mrs a Mr.

````python
# first with fixed names — no input

def greet(name):
    name = name.lower().strip()
    # Order matters — 'mrs' must be checked first, otherwise 'mr' catches it
    if name.startswith('mrs'):
        return 'Dear mam, welcome to our site.'
    elif name.startswith('mr'):
        return 'Dear Mr., welcome to our site.'
    elif name.startswith('md'):
        return 'Dear sir, welcome to our site.'
    else:
        return 'Dear sir/mam, welcome to our site.'


for n in ['Md Wabidur', 'Mr Rahim', 'Mrs Sadia', 'Karim']:
    print(f'{n:<12} → {greet(n)}')
````

# Part 4 — Classes / Objects

In Python **almost everything is an object** — each with its own **data (properties)** and
**actions (methods)**. Until now we have been using ready-made objects (list, string, DataFrame).
Today we learn to build **our own objects**.

A **class** is the mould or blueprint from which objects are made — like the plan of a house.
The actual houses built from that plan are the **objects**.

````python
# the simplest class — some data and one method

class MyClass:
    name = 'Md Wabidur Rahman'
    profession = 'Researcher'

    def mean(numbers):
        add = 0
        for i in numbers:
            add = add + i
        return add / len(numbers)


# using the class data
print('name       :', MyClass.name)
print('profession :', MyClass.profession)

# using the class method
li = [1, 2, 3, 4, 5, 5, 6, 7, 7, 9]
print('mean       :', MyClass.mean(li))
````

````python
# checking the same mean against three familiar modules

import statistics as st
import numpy as np

print('our class    :', MyClass.mean(li))
print('statistics   :', st.mean(li))
print('numpy        :', np.mean(li))
````

### `__init__()` and `self` — a real class

In the class above, every object would have the same name. But in reality each object is different —
one car is black, another is white. That is what **`__init__()`** is for.

- **`__init__()`** — runs **automatically** at the moment the object is created, setting the initial values
- **`self`** — "this object itself"; it must be the first parameter of every method

````python
# a car class with __init__() and self

class Car:
    def __init__(self, name, color):
        self.name = name        # this object's name
        self.color = color      # this object's colour

    def start(self):
        return f'The engine of the {self.color} {self.name} has started.'


# two separate objects made from the blueprint
car1 = Car('Allion', 'Black')
car2 = Car('Premio', 'White')

print(car1.name, '-', car1.color)
print(car2.name, '-', car2.color)
print()
print(car1.start())
print(car2.start())
````

### A practical example — a student class

The grading job from Class 04, now inside a class. Each student is an object,
with their own name and marks, and their own methods for the average and the result.

````python
# Student class — its own data, its own calculations

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

    def result(self):
        if self.average() >= 33:
            return 'Pass'
        else:
            return 'Fail'


s1 = Student('Rahim', [82, 75, 91])
s2 = Student('Karim', [28, 30, 25])       # mean below 33 -> Fail

for s in [s1, s2]:
    print(f'{s.name:<8} mean {s.average():.2f}  ->  {s.result()}')
````

# Part 5 — Modules

A **module** is a file of someone else's code — full of functions and classes, which we pull into
our own code with `import`. There is no need to reinvent the wheel.

All day today we have been using three modules — `pandas`, `numpy` and `statistics`.
Now let us set out the rules properly.

### Three ways to import

````python
# 1. the whole module — you write module.function() every time

import math
print('math.sqrt(16) :', math.sqrt(16))
print('math.pi       :', math.pi)

# 2. a short name (alias) — so the long name is not repeated
import numpy as np
print('np.mean       :', np.mean([1, 2, 3, 4]))

# 3. only what you need from the module
from statistics import mean, stdev
print('mean          :', mean([1, 2, 3, 4]))
````

````python
# math — mathematical functions

import math
print('math.factorial(5) :', math.factorial(5))
print('math.ceil(4.2)    :', math.ceil(4.2))     # round up
print('math.floor(4.8)   :', math.floor(4.8))    # round down
````

# Part 6 — Everything together: analysing `sample1.csv`

All five of today's topics in one place — reading the data with a **module**, computing with a
**function**, sorting with a **lambda**, and building the report with **string methods**.

````python
# importing modules and reading the data

import pandas as pd

df = pd.read_csv('sample1.csv')
df.columns = [c.replace('\n', ' ').strip() for c in df.columns]

print('columns :', list(df.columns))
df.head()
````

````python
# a function — summarising the probability

def summarise(series):
    return {
        'total years' : len(series),
        'maximum'     : round(series.max(), 3),
        'minimum'     : round(series.min(), 3),
        'mean'        : round(series.mean(), 3),
    }

for key, value in summarise(df.Probability).items():
    print(f'{key:<10}: {value}')
````

````python
# lambda — ranking the years by probability (highest first)

pairs = list(zip(df.Year, df.Probability))
top5 = sorted(pairs, key=lambda pair: pair[1], reverse=True)[:5]

print('the five years with the highest probability:')
for year, prob in top5:
    print(f'  {year} → {prob}')
````

````python
# string methods — building a clean line of report

wettest_year = int(top5[0][0])
wettest_prob = top5[0][1]

report = f'The highest probability was in {wettest_year} ({wettest_prob}).'
print(report)
print('in upper case :', report.upper() if report.isascii() else report)
print('word count    :', len(report.split()))
````
