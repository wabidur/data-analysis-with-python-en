# Part 1 — List


## What is a List?

- A List is the most widely used container for holding several values together.
- It is written inside square brackets `[ ]`, with the items separated by commas.
- It is **ordered**, **changeable**, and **duplicate values are allowed**.
- A single list can hold different data types — text, integers, decimals, all together.

### Index — counting starts at zero

```
list_1 = ['Bangladesh', 'India', 'Pakistan', 'Nepal', 18, 'July', 2022, 9.08]
index:        0          1          2          3       4     5      6      7
negative:    -8         -7         -6         -5      -4    -3     -2     -1
```

A **negative index** counts from the end — `li[-1]` means the very last element.
You do not need to write `li[len(li)-1]` to reach the last item; `li[-1]` is enough.

````python
# Creating a List and getting items by index

list_1 = ['Bangladesh', 'India', 'Pakistan', 'Nepal', 18, 'July', 2022, 9.08]

print(list_1)
print(type(list_1))
print()

first_content = list_1[0]      # the index must be a whole number
fifth_content = list_1[4]

print('First content  :', first_content)
print('Fifth content  :', fifth_content)      
print()

print('Length of the list :', len(list_1))
print('Last content            :', list_1[len(list_1) - 1])
print('Last content (negative) :', list_1[-1])
print('First content (negative):', list_1[-8])
````

````python
print(list_1[-2])
````

### Slicing — several items at once

`li[start : end]` — **the start position is included, the end position is excluded.**

That "the last one is left out" rule is what confuses beginners most.
`li[1:4]` means positions 1, 2 and 3 — not position 4.

````python
list_1 = ['Bangladesh', 'India', 'Pakistan', 'Nepal', 18, 'July', 2022, 9.08]

print(list_1[1:4])    # positions 1 to 3 (4 excluded)
print(list_1[2:])     # from position 2 all the way to the end
print(list_1[:6])     # from the start to position 5 (6 excluded)
print(list_1[:])      # the whole list — start to end
print(list_1[-3:])    # the last three
````

### Changing a List

A List is **changeable** — you can modify the values after it has been created.
You can change one position, or several positions at once.

````python
list_1 = ['Bangladesh', 'India', 'Pakistan', 'Nepal', 18, 'July', 2022, 9.08]

list_1[1] = 'Srilanka'    # one position
#list_1[1] = 'UK'
list_1[2:4] = ['America', 'Canada']         # two positions at once
#list_1[2:4] = ['Brazil', 'Arzentina']
print(list_1)
print('Length:', len(list_1))
````

### Three ways to add an item

| Method | What it does | Argument |
|--------|-----|----------|
| `append(x)` | adds one item **at the end** | one |
| `insert(i, x)` | places an item **at a given position** | two (position, value) |
| `extend(li)` | appends **every item of another list** | a list |

> **Remember the difference:** if you `append` a list, it is placed as **a single item**
> (a list inside a list). But `extend` spreads the inner items out separately.

````python
list_1 = ['Bangladesh', 'Srilanka', 'America', 'Canada']

list_1.append('abc')              # add at the end
print('append  :', list_1, '| length:', len(list_1))

list_1.insert(2, 'China')         # insert at position 2
print('insert  :', list_1, '| length:', len(list_1))
````

````python
#list_1.insert(2, 'China')         # insert at position 2
#print('insert  :', list_1, '| length:', len(list_1))
````

````python
# extend — appending another list

li_1 = ['My', 'name', 'is', 'Md Wabidur Rahman.']
li_2 = ['Class', 'time:', 22, 'September', 2026]

li_2.extend(li_1)
print(li_2)
````

````python
# append versus extend — see the difference

a = [1, 2]
b = [1, 2]

a.append([3, 4])      # placed the whole list as a single item
b.extend([3, 4])      # spread the inner items out

print('append :', a, '| length:', len(a))
print('extend :', b, '| length:', len(b))
````

### join() — turning a list into a sentence

`' '.join(li)` joins the items of a list into one string.
The `' '` in front is the **separator** — it decides what goes between the pieces.

⚠ **Condition:** every item of the list must be a string. If a number is present you get a `TypeError`.

````python
li_4 = ['My', 'name', 'is', 'Md Wabidur Rahman.']

print(' '.join(li_4))        # with a space
print('-'.join(li_4))        # with a hyphen
print(', '.join(li_4))       # with a comma

# if there are numbers, convert with str() first:
nums = [2024, 2025, 2026]
print(', '.join(str(n) for n in nums))
````

### sort() — ordering

`sort()` reorders the original list in place.
Pass `reverse = True` to sort in descending order (largest to smallest).

⚠ **One surprising detail:** when sorting text, Python looks at the character codes —
so **digits come first, then uppercase letters, and lowercase letters last**.
That is why `'Mango'` (capital M) comes before `'banana'` (lowercase b).

````python
li_6 = ["orange", "Mango", "kiwi", "pineapple", "banana", "3"]
li_7 = [100, 50, 65, 82, 23]

print('before :', li_6)
li_6.sort()
print('after  :', li_6)     # '3' first, 'Mango' (capital M) before 'banana'

li_7.sort()
print('numbers :', li_7)
````

````python
# sorting in reverse order

li_6.sort(reverse=True)
print(li_6)

li_7.sort(reverse=True)
print(li_7)
````

````python
# to sort while ignoring upper/lower case

li_8 = ["orange", "Mango", "kiwi", "Pineapple", "banana"]

li_8.sort(key=str.lower)     # treats everything as lowercase while sorting
print(li_8)
````

## ⚠ Very important: copy() versus `=`

In the last class we learned the `is` operator — remember? This is its biggest application.

Writing `b = a` does **not create a new list** — it only attaches a second label to the same list.
So anything you change through `b` also shows up in `a`!

For a real copy you must use **`a.copy()`**.

````python
#
a = [1, 2, 3]
b = a                # not a copy — a second name for the same list

b.append(4)

print('a :', a)      # a changed too!
print('b :', b)
print('a is b :', a is b)      # True — the same object
````

````python
# the correct way — copy()

a = [1, 2, 3]
c = a.copy()         # a real copy

c.append(4)

print('a :', a)      # a is unchanged
print('c :', c)
print('a is c :', a is c)      # False — different objects
print('a == c :', a == c)      # False — the values now differ too
````

````python
#
a = [1, 2, 3]
b = a.copy()                # a copy this time

b.append(4)

print('a :', a)      # a is unchanged
print('b :', b)
print('a is b :', a is b)      # False — different objects
````

### Unpacking — splitting into several variables at once

The number of items in the list and the number of variables must be **equal**, otherwise you get a `ValueError`.

````python
li = [1, 2, 3, 4]

n1, n2, n3, n4 = li          # four items, four variables

print(n1, n2, n3, n4)
print('n2 + n3 =', n2 + n3)
````

# Part 2 — Tuple


## What is a Tuple?

- A Tuple also holds several values, inside round brackets `( )`.
- It is **ordered** — indexing and slicing work just as they do for a List.
- But it is **unchangeable / immutable** — once created, the values inside cannot be changed.

**So why do we need Tuples?**
1. **Safety** — information that should not change (the seven days of the week, a coordinate) is safe from being modified by accident.
2. **Speed** — a Tuple is slightly faster and uses less memory.
3. **It can be a dictionary key** — a List cannot, a Tuple can.

````python
tpl = (1, 3, 4, 5, 6)

print(tpl)
print(type(tpl))
print('Length:', len(tpl))
````

### You get a Tuple even without brackets

Commas alone are enough for Python to treat it as a Tuple — the brackets are optional.

````python
tpl_1 = 1, 3, 4, 'abc'        # a Tuple without brackets

print(tpl_1)
print(type(tpl_1))
print('Length:', len(tpl_1))
````

### ⚠ A one-item Tuple — do not forget the comma

To write a Tuple with a single item you **must** put a comma at the end.
Without the comma Python reads it as an ordinary pair of brackets, not a Tuple.

This is a very common beginner's mistake.

````python
wrong = (5)        # no comma -> this is a number, not a Tuple
right = (5,)       # comma present -> a Tuple

print(wrong, type(wrong))
print(right, type(right))
````

### Indexing and slicing — just like a List

````python
tpl_1 = (1, 3, 4, 'abc')

print(tpl_1[0])       # the first item
print(tpl_1[-1])      # the last item
print(tpl_1[1:3])     # positions 1 and 2
````

### A Tuple cannot be changed — proof

Uncomment the line below and running it raises a `TypeError`.
That is the defining property of a Tuple.

````python
tpl_1 = (1, 3, 4, 'abc')

#tpl_1[0] = 100        # uncomment to see it -> TypeError

print('Tuple unchanged:', tpl_1)
````

### If you really must change it — convert to a List

The trick is simple: Tuple → List → change → Tuple again.

````python
tpl_1 = (1, 3, 4, 'abc')

li_10 = list(tpl_1)          # Tuple -> List
print('as a List :', li_10)

li_10.append(16)             # now it can be changed
li_10[3] = 'Bangladesh'
print('after the change :', li_10)

tpl_3 = tuple(li_10)         # back to a Tuple
print('Tuple again :', tpl_3)
print(type(tpl_3))
````

### Tuple unpacking


````python
person = ('Wabidur', 'Dhaka', 2026)

name, city, year = person

print('Name :', name)
print('City :', city)
print('Year :', year)
````

### Deleting a whole Tuple

You cannot delete items, but you can delete the whole Tuple with `del`.

# Part 3 — Set


## What is a Set?

- A Set holds several values inside curly brackets `{ }`.
- It is **unordered** — there is no fixed sequence. That is why **you cannot look items up by index**.
- It holds **no duplicates** — give the same value twice and it is kept once.
- Items cannot be changed, but items can be **added** and **removed**.

> **⚠ Watch out in class:** because a Set is unordered, printing it may show the items
> in a **different order each time**. If your screen does not match mine,
> that is not a mistake — it is normal Set behaviour.

````python
# notice — 'Friday' is written twice

set_1 = {'Friday', 'Saturday', 'Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Friday'}

print(set_1)
print(type(set_1))
print('Length:', len(set_1))      # 6, not 7 — the duplicate was dropped
````

### List to Set — the easiest way to remove duplicates

This is the most widely used practical application of a Set.

````python
li = ['Friday', 'Saturday', 'Sunday', 'Friday', 'Monday', 'Saturday']

set_2 = set(li)              # List -> Set : the duplicates are gone

print('original List :', li, '| length:', len(li))
print('as a Set      :', set_2, '| length:', len(set_2))

back_to_list = list(set_2)   # back to a List
print('List again    :', back_to_list)
````

### Indexing does not work on a Set — proof

````python
set_1 = {'Friday', 'Saturday', 'Sunday'}

#print(set_1[0])      # uncomment for a TypeError — a Set has no index

# to search, use `in` from Class 02:
print('Is Friday there?', 'Friday' in set_1)
print('Is Monday there?', 'Monday' in set_1)
````

### Adding items — add() and update()

| Method | What it does |
|--------|-----|
| `add(x)` | adds **one** item |
| `update(s)` | adds **every item** of another set or list |

````python
set_1 = {'Friday', 'Saturday', 'Sunday', 'Monday', 'Tuesday', 'Wednesday'}

set_1.add('Thursday')             # add one
print(set_1)
print('Length:', len(set_1))
````

````python
fruits = {"apple", "banana", "cherry"}

set_1.update(fruits)              # add them all
print(set_1)
print('Length:', len(set_1))
````

### Removing items — remove() versus discard()

| Method | What it does | If the value is absent |
|--------|-----|---------------|
| `remove(x)` | removes it | raises **`KeyError`** |
| `discard(x)` | removes it | **does nothing, silently** |
| `pop()` | removes an arbitrary item and returns it | — |
| `clear()` | empties the whole Set | — |

> **Which should you use?** If you are not certain the value is present, `discard()` is the safe choice.

````python
set_1.remove('apple')         # present, so it was removed
print(set_1)

set_1.discard('banana')       # present, removed
print(set_1)

set_1.discard('mango')        # absent, yet no error
print('discard() is safe — no error raised')

#set_1.remove('mango')       # uncomment for a KeyError
````

### pop() — on a Set you cannot predict which item goes

A Set is unordered, so there is no way to know in advance which item `pop()` will remove.
On a List, `pop()` always removes the last item — on a Set it does not.

````python
set_3 = {'Friday', 'Saturday', 'Sunday', 'Monday'}

removed = set_3.pop()

print('removed   :', removed)
print('remaining :', set_3)
````

````python
# copy() and clear()

set_4 = set_3.copy()
print('copy :', set_4)

set_4.clear()
print('after clear :', set_4)     # set() — an empty Set
print(type(set_4))
````

## Set arithmetic

This is where the real power of a Set lies. Each operation has **two forms** — a function and a symbol.

| Operation | Function | Symbol | Meaning |
|-----|----------|-------|------|
| Union | `a.union(b)` | `a \| b` | **every** item from both |
| Intersection | `a.intersection(b)` | `a & b` | the items present **in both** |
| Difference | `a.difference(b)` | `a - b` | in a but not in b |
| Symmetric difference | `a.symmetric_difference(b)` | `a ^ b` | **the items not shared by both** |

````python
set_1 = {'Friday', 'Saturday', 'Sunday', 'Monday', 'Tuesday', 'Wednesday'}
set_2 = {'Friday', 'Saturday', 'Sunday', 'apple', 'banana', 'cherry'}

print('set_1 :', set_1)
print('set_2 :', set_2)
````

````python
# Union — every item from both

print('function :', set_1.union(set_2))
print('symbol   :', set_1 | set_2)
````

````python
# Intersection — the items present in both

print('function :', set_1.intersection(set_2))
print('symbol   :', set_1 & set_2)
````

````python
# Difference — in set_1 but not in set_2

print('function :', set_1.difference(set_2))
print('symbol   :', set_1 - set_2)

# reversing the order changes the result:
print('the other way :', set_2 - set_1)
````

````python
# Symmetric difference — the items not shared by both

x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

print('function :', x.symmetric_difference(y))
print('symbol   :', x ^ y)
print('(apple is in both, so it was dropped)')
````

### The `_update` forms — they modify the original Set

`intersection()` returns a new Set, but `intersection_update()`
modifies the original Set in place and returns nothing.

````python
a = {'Friday', 'Saturday', 'Sunday', 'Monday'}
b = {'Friday', 'Saturday', 'apple'}

new_set = a.intersection(b)       # returns a new Set, a is unchanged
print('new_set :', new_set)
print('a       :', a, '  <- unchanged')

a.intersection_update(b)          # changes a itself
print('a after update :', a)
````

````python
# notice which one update() changes

p = {'Friday', 'Saturday'}
q = {'apple', 'banana'}

p.update(q)

print('p :', p, '  <- p has changed')
print('q :', q, '  <- q is unchanged')
````

# Part 4 — Dictionary


## What is a Dictionary?

- A Dictionary stores information in **key : value** pairs.
- It is written inside curly brackets `{ }`, with a colon `:` separating key from value.
- It is **ordered** (since Python 3.7), **changeable**, and **keys are unique**.
- A value can be of any data type — a number, text, even a list or another dictionary.

**An analogy:** just like a real dictionary — look up a word (key) and you get its meaning (value).
In a List you search by position number; in a Dictionary you search **by name**.

````python
dict_1 = {'Akib': 40, 'Saif Hossain': 100, 'Satil': 100, 'Antor': 60}
dict_2 = {'brand': 'Ford', 'model': 'Mustang', 'year': 2022}
dict_3 = {1: 'A', 2: 'B', 3: 'C'}

print(dict_1)
print(dict_2)
print(dict_3)
print(type(dict_1))
````

### keys(), values(), items()

| Method | What it gives |
|--------|---------|
| `keys()` | all the keys |
| `values()` | all the values |
| `items()` | keys and values, in pairs |

````python
print('keys   :', dict_1.keys())
print('values :', dict_1.values())
print('items  :', dict_1.items())
````

````python
# looping over a Dictionary — items() is the most used

for name, marks in dict_1.items():
    print(name, '->', marks)
````

### Getting information out — `[ ]` versus `get()`

| Way | If the key is absent |
|-------|---------------|
| `d['key']` | raises **`KeyError`** |
| `d.get('key')` | returns **`None`**, no error |
| `d.get('key', 'default')` | returns the default you chose |

> **Which should you use?** Use `[ ]` when the key is certain, `get()` when it is not.

````python
dict_2 = {'brand': 'Ford', 'model': 'Mustang', 'year': 2022}

print(dict_2['year'])            # present
print(dict_2.get('model'))       # present

print(dict_2.get('color'))       # absent -> None, no error
print(dict_2.get('color', 'no data'))    # your own default value

#print(dict_2['color'])         # uncomment for a KeyError
````

### Adding and changing

In a Dictionary, a new key **adds** an entry and an existing key **replaces** one —
and both are written exactly the same way. So you have to be careful.

````python
dict_2 = {'brand': 'Ford', 'model': 'Mustang', 'year': 2022}

dict_2['model'] = 'Shelby'                 # existing key -> replaced
print(dict_2)

dict_2['name'] = 'Md Wabidur Rahman'       # new key -> added
print(dict_2)

dict_2.update({'year': 2026})              # update() can change it too
print(dict_2)
````

### Removing

| Method | What it does |
|--------|-----|
| `pop('key')` | removes that key and returns its value |
| `popitem()` | removes the **last inserted** pair |
| `del d['key']` | removes that key |
| `clear()` | empties the whole dictionary |

````python
removed = dict_2.pop('model')
print('removed :', removed)
print(dict_2)

dict_2.popitem()               # removed the last one
print('after popitem :', dict_2)
````

````python
dict_copy = dict_2.copy()      # copy() — the same rule applies here

dict_copy.clear()
print(dict_copy)
print(type(dict_copy))
````

### Putting a list or a tuple in the value

The value of a Dictionary can be anything — even a whole list.
This structure maps directly onto a pandas table.

````python
dict_4 = {'a': (77, 88, 99), 'b': [33, 53, 65], 'c': [44, 24, 54]}

print(dict_4)
print()
print("dict_4['a'] =", dict_4['a'])
print("dict_4['b'][1] =", dict_4['b'][1])      # position 1 of b
````

````python
# Dictionary -> pandas DataFrame : the keys become column names

import pandas as pd

df_dict = pd.DataFrame(dict_4)
print(df_dict)
````

### Nested dictionary — a dictionary inside a dictionary

Real information is often arranged in layers. That is when this structure is useful.

````python
training = {
    'UniV': {
        'course name': 'Python',
        'participants': 20
    },
    'Personal Course': {
        'course name': 'GIS and Remote Sensing',
        'participants': 'All'
    }
}

print(training)
print()
print('outer keys :', list(training.keys()))
print("going inside :", training['UniV']['course name'])
````

### Building one with the dict() function

With `dict()` the keys are written as **bare names**, without quotes.

````python
dict_5 = dict(name='Wabidur', city='Dhaka', year=2026)

print(dict_5)
print(type(dict_5))
````

## ⚠ Important: in `dict(name=value)` the name itself becomes the key, not the variable

To build a dictionary from two lists, the correct tool is **`zip()`**.

````python
years = [1960, 1961, 1962, 1963]
rainfall = [130, 84, 76, 89]

# ✗ the wrong way
wrong = dict(years=rainfall)
print('wrong  :', wrong)
print('keys   :', list(wrong.keys()), '  <- the text, not the variable!')
print()

# ✓ the correct way — zip()
correct = dict(zip(years, rainfall))
print('correct :', correct)
print('rainfall in 1962 :', correct[1962])
````

Now we take a real dataset and apply all four of today's containers to it.
The file `sample1.csv` holds annual rainfall and its probability from 1960 to 1981.

````python
import pandas as pd

df = pd.read_csv('sample1.csv')
print(df.head())
print()
print('total rows :', len(df))
print('column names :', list(df.columns))
````

````python
# a pandas column -> List and Tuple

year_list = list(df.Year)
rainfall_list = list(df["Annual Rainfal\n(cm)"])

year_tuple = tuple(df.Year)          

print('List  :', year_list[:8], '...')
print('Tuple :', year_tuple[:8], '...')
print()
print('Type  :', type(year_list), type(year_tuple))
````

### Looking information up by year, with a List and with a Dictionary

Compare the two approaches side by side — which is simpler?

````python
# Way 1 — List + index() : find the position first, then the value there

target_year = 1970

serial = year_list.index(target_year)       # which position holds it
print('position :', serial)
print('rainfall in', target_year, ':', rainfall_list[serial], 'cm')
````

````python
# Way 2 — Dictionary : straight to the name

rain_dict = dict(zip(year_list, rainfall_list))

print(rain_dict[1970], 'cm')          # no position lookup needed
print(rain_dict[1977], 'cm')
print()
print('Is 1970 in the data?', 1970 in rain_dict)
print('Is 1990 in the data?', 1990 in rain_dict)
````

**This is the main advantage of a Dictionary** — with a List you first have to find the position, with a Dictionary you go straight to the name.
As the data grows, that difference matters for speed too.

### Finding unique values with a Set

````python
prob_list = list(df.Probability)

print('total values   :', len(prob_list))
print('unique values  :', len(set(prob_list)))
print('Any duplicates?', len(prob_list) != len(set(prob_list)))
````

````python
# classify the rainfall and find the unique classes

categories = []

for r in rainfall_list:
    if r >= 120:
        categories.append('Heavy')
    elif r >= 90:
        categories.append('Medium')
    else:
        categories.append('Low')

print('class of each year :', categories[:8], '...')
print('how many classes   :', set(categories))
````

````python
# how many years in each class — counting with a Dictionary

count = {}

for c in categories:
    count[c] = count.get(c, 0) + 1      # the default value of get() is useful here

print(count)
````

### Sorting, and the maximum and minimum

````python
sorted_rain = sorted(rainfall_list)      # sorted() returns a new list, the original is unchanged

print('lowest 5  :', sorted_rain[:5])
print('highest 5 :', sorted_rain[-5:])
print()
print('maximum rainfall :', max(rainfall_list), 'cm')
print('minimum rainfall :', min(rainfall_list), 'cm')
print('mean rainfall    :', round(sum(rainfall_list) / len(rainfall_list), 2), 'cm')
````

````python
# which year had the highest rainfall?

max_rain = max(rainfall_list)
max_index = rainfall_list.index(max_rain)

print('The highest rainfall,', max_rain, 'cm, fell in', year_list[max_index])
````

|---|---|---|---|---|

### Five things worth remembering from today

1. **`b = a` is not a copy** — it is a second name for the same object. For a real copy use `a.copy()`.
2. **A one-item Tuple needs a comma** — `(5,)`, not just `(5)`.
3. **Use `set()` to drop duplicates** — it is a one-liner.
4. **In `dict(name=value)` the name becomes the key**, not the variable. Use `zip()` to join two lists.
5. **`get()` is safe** — it raises no error when the key is missing, and it accepts a default value.
