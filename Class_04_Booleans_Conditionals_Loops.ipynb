# Part 1 — Boolean

To a computer the world is very simple — it is either **yes** or **no**.
In Python those two are `True` and `False`. Their type is called **`bool`**.

Three things to remember —
- The first letter is **capital**: `True`, `False` (writing `true` gives a `NameError`)
- **No quotation marks**: `'True'` is text, `True` is a bool
- Underneath, `True` = **1** and `False` = **0**

````python
# Meeting the Boolean

x = True
y = False

print('x =', x, '| type:', type(x))
print('y =', y, '| type:', type(y))
````

````python
# put it in quotes and it is no longer a bool — it becomes text

real_bool = True
fake_bool = 'True'

print(real_bool, type(real_bool))
print(fake_bool, type(fake_bool))
print()
print('Are the two equal? :', real_bool == fake_bool)
````

### A Boolean really is a number — `True` = 1, `False` = 0

This is exactly why `sum()` can count "how many conditions were met". In Class 02 we wrote
`hot.sum()` in pandas to count the hot days — this is the mechanism behind it.

````python
# True and False can be added up

print('True + True   =', True + True)
print('True + False  =', True + False)
print('True * 10     =', True * 10)
print()

results = [True, False, True, True, False, True]
print('how many True  :', sum(results))
print('how many False :', len(results) - sum(results))
````

### Booleans are produced by comparisons

Every **comparison operator** from Class 02 returns a `bool`.

````python
# the result of every comparison is a bool

a = 10
b = 7

print('a > b   →', a > b)
print('a < b   →', a < b)
print('a == b  →', a == b)
print('a != b  →', a != b)
print()
print('type of the result :', type(a > b))
````

### `bool()` — turning anything into a yes/no

In Python some values count as **"no"** all by themselves. They are called **falsy** values —
`0`, the empty string `''`, the empty list `[]`, the empty dictionary `{}`, and `None`.
Almost everything else is **truthy**, that is, "yes".

````python
# which are falsy, which are truthy

items = [0, 1, -5, '', 'a', [], [1, 2], {}, {'a': 1}, None, 0.0, 3.14]

print('value           ->  bool()')
print('-' * 30)
for item in items:
    print(f'{repr(item):<15} →  {bool(item)}')
````

> **Why is this useful?** Because of this rule, writing `if my_list:` already tells you
> whether the list is empty — you do not need `if len(my_list) > 0:`.

````python
# an empty list says "no" all by itself

basket = []

if basket:
    print('the basket has something in it')
else:
    print('the basket is empty')

basket.append('Mango')

if basket:
    print('the basket now holds :', basket)
````

# Part 2 — if … else Statement

`if` alone says — **if the condition is true**, do this; if it is false, do nothing.
But what if you also want to do something when it is false? That is what **`else`** is for.

```
if (condition):
    ← this block runs when the condition is true
else:
    ← this block runs when the condition is false
```

**Two things you must not forget —**
1. A **colon `:`** at the end of the `if` and `else` lines
2. The lines inside must be **indented** — usually four spaces

````python
# the simplest example

a = 5
c = 3

if 3 == a:
    print(a)
else:
    print(c)
````

> Above, `3 == a` asks "is 3 equal to a?" — a is 5, so the answer is `False`,
> which is why the `else` branch ran and `3` was printed.

````python
# which of two numbers is larger

num_1 = 23
num_2 = 50

if num_1 > num_2:
    print(num_1, 'is greater than', num_2)
else:
    print(num_2, 'is greater than', num_1)
````

### Indentation — Python's strictest rule

Other languages mark a block with `{ }`. In Python, **whitespace** defines the block.
If the inner lines are not indented you get an `IndentationError`.

````python
# let us make the mistake on purpose and see the message

code_with_error = '''
if 5 > 3:
print('not indented')
'''

try:
    exec(code_with_error)
except IndentationError as e:
    print('❌ IndentationError:', e.msg)
    print('   Fix: the print line has to be indented by four spaces.')
````

# Part 3 — if … elif … else Statement

`else` gives you only **two** paths. But grading needs seven —
A+, A, A−, B, C, D, F. That is when you need **`elif`** (short for else + if).

```
if  condition-1 :    ← if true this runs, and the rest are never looked at
elif condition-2 :   ← tested only when the one above was false
elif condition-3 :
else :               ← when none of the above matched
```

**The most important rule:** Python checks **from top to bottom**, and
**runs the first condition it finds true, then leaves** — it never looks at the rest.
So the **order** of the conditions matters.

````python
# working out the grade — with a fixed value

mark = 67

if mark >= 80:
    grade = 'A+'
elif mark >= 70:
    grade = 'A'
elif mark >= 60:
    grade = 'A-'
elif mark >= 50:
    grade = 'B'
elif mark >= 40:
    grade = 'C'
elif mark >= 33:
    grade = 'D'
else:
    grade = 'F'

print('mark  :', mark)
print('grade :', grade)
````

### What happens when the order is wrong — worth seeing once

In the code above, the **largest condition came first**. If we reverse it and put the
**smallest condition first**, then even a score of 95 satisfies `mark >= 33` immediately,
and the result becomes **D**!

````python
# the wrong order — the smallest condition first

mark = 95

if mark >= 33:
    wrong_grade = 'D'
elif mark >= 80:
    wrong_grade = 'A+'
else:
    wrong_grade = 'F'

print('grade in the wrong order :', wrong_grade, '  <- D for a score of 95!')

````

````python
# the correct order — the largest condition first
if mark >= 80:
    right_grade = 'A+'
elif mark >= 33:
    right_grade = 'D'
else:
    right_grade = 'F'

print('grade in the correct order :', right_grade)
````

### Putting the grading in a function — write once, use many times

You will not have to write the same seven conditions again. The function below will also
be useful inside the loop in Part 5.

````python
def get_grade(mark):
    """Take a mark and return the grade."""
    if mark >= 80:
        return 'A+'
    elif mark >= 70:
        return 'A'
    elif mark >= 60:
        return 'A-'
    elif mark >= 50:
        return 'B'
    elif mark >= 40:
        return 'C'
    elif mark >= 33:
        return 'D'
    else:
        return 'F'


# testing the boundary values — this is where mistakes show up
for m in [100, 80, 79, 70, 60, 50, 40, 33, 32, 0]:
    print(f'{m:>4} → {get_grade(m)}')
````

# Part 4 — Nested if (a condition within a condition)

A whole `if…elif…else` can be placed **inside** an `if` block. This is called
**nesting**. The inner condition is tested only when the outer one is true.

````python
# positive, negative or zero — with a nested if

for num in [7, -67, 0]:

    if num >= 0:
        if num == 0:
            result = 'Zero'
        else:
            result = 'Positive number'
    else:
        result = 'Negative number'

    print(f'{num:>4} → {result}')
````

### Leap year — the nicest example of a nested if

**The rule:**
- If a year is divisible by **4** → it is a leap year
- **But** if it is also divisible by **100** → then it must be divisible by **400** as well
- So a year divisible by 100 but not by 400 → is **not** a leap year

For this we need the **modulus `%`** operator, which returns the remainder.
A remainder of **0** means it divides exactly.

````python
# first let us understand the % operator

print('2024 % 4   =', 2024 % 4,   '-> remainder 0, so divisible by 4')
print('2024 % 100 =', 2024 % 100, '-> there is a remainder, so not divisible by 100')
print('2000 % 400 =', 2000 % 400, '-> remainder 0, so divisible by 400')
print('1900 % 400 =', 1900 % 400, '-> there is a remainder, so not divisible by 400')
````

````python
# Leap Year — nested if

def is_leap(year):
    if year % 4 == 0:
        if year % 100 == 0:
            if year % 400 == 0:
                return f'{year} is Leap Year'
            else:
                return f'{year} is not Leap Year'
        else:
            return f'{year} is Leap Year'
    else:
        return f'{year} is not Leap Year'


# testing with four different kinds of year
for y in [2024, 2023, 2000, 1900]:
    print(is_leap(y))
````

> **Notice:** `2000` is a leap year **but** `1900` is not — both are divisible by 100,
> and the only difference is the 400 condition. This single example shows why a nested if is needed.

````python
# the same job with logical operators — in a single line (and / or from Class 02)

def is_leap_short(year):
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)


print('year   nested    and/or    do they agree?')
print('-' * 44)
for y in [2024, 2023, 2000, 1900, 1996, 2100]:
    nested = 'is not' not in is_leap(y)
    short = is_leap_short(y)
    print(f'{y}   {str(nested):<9} {str(short):<9} {nested == short}')
````

# Part 5 — For Loop

### Where is the problem?

To print your name ten times you would write `print()` ten times. And a hundred times?
This is exactly where a **loop** is needed.

A `for` loop takes each element of a **sequence** (List, Tuple, Set, Dictionary, String)
in turn and runs the body once for each.

````python
# the painful way — writing the same line ten times (only three shown)

print('My name is Md Wabidur Rahman')
print('My name is Md Wabidur Rahman')
print('My name is Md Wabidur Rahman')
````

````python
# the easy way — ten times with a for loop

for i in range(10):
    print('My name is Md Wabidur Rahman')
````

### `range()` — a staircase of numbers

`range(10)` means **0 to 9** — ten numbers in all. It starts at **0**,
and the **last number is excluded** (exactly like the slicing in Class 03).

````python
# what range() actually holds

print('range(10) →', list(range(10)))
print('range(1, 5) →', list(range(1, 5)))          # 1 to 4
print('range(0, 10, 2) →', list(range(0, 10, 2)))  # in steps of 2
print('range(10, 0, -1) →', list(range(10, 0, -1)))  # in reverse
````

````python
# when you only need to count

for i in range(10):
    print(i)
````

### Looping over a string — one character at a time

`for` works not only on a List but on any **sequence**. A string is itself a
sequence, so each character comes through one by one.

````python
# each character of the name

name = 'Md Wabidur Rahman'

for ch in name:
    print(ch)
````

````python
# putting it to use — counting the vowels

name = 'Md Wabidur Rahman'
vowels = 'aeiouAEIOU'
count = 0

for ch in name:
    if ch in vowels:
        count = count + 1

print('total characters :', len(name))
print('vowels           :', count)
````

### The accumulator technique

Start a variable at **zero** outside the loop, then add to it each time round —
this is one of the most widely used techniques in all of programming.

````python
# the sum from 1 to 4

result = 0

for i in range(1, 5):
    print('now i =', i, '| result before adding =', result)
    result = result + i

print()
print('total sum :', result)
````

### Finding the largest number

Start by assuming the first element is "the largest so far", then compare each of the
others against it to see whether anything bigger turns up.

````python
# the largest number without max()

numbers = [3, 4, 6, 2, 54, 6, 3, 5, 7, 5, 3, 57, 5, 34]
max_n = numbers[0]

for n in numbers:
    if n > max_n:
        max_n = n

print('the largest number :', max_n)
print('checking with max() :', max(numbers))
````

### Nested for loops — a loop inside a loop

For each single turn of the outer loop, the inner loop runs **all the way through**.
So the total number of runs is **3 × 3 × 2 = 18**.

````python
# a three-level nested loop

li_1 = ['apple', 'banana', 'cherry']
li_2 = ['mango', 'jackfruit', 'pineapple']
li_3 = ['a', 'b']

total = 0

for i in li_1:
    for j in li_2:
        for k in li_3:
            print('i=', i, '& j=', j, '& k=', k)
            total = total + 1

print()
print('total number of runs :', total, '  (3 x 3 x 2)')
````

> **A warning:** the more levels you add, the work grows **multiplicatively**. With three
> lists of 100 items each it would run a million times. Think twice before using a nested
> loop on large data.

# Part 6 — `break` and `continue`

| Keyword | What it does | How to remember it |
|------|-----|------------------|
| `break` | **stops** the whole loop and leaves | break out |
| `continue` | **skips** this step and moves to the next | carry on, skip this one |

````python
# break — once we have found what we wanted there is no need to keep going

li_2 = ['mango', 'jackfruit', 'pineapple']

for i in li_2:
    if i == 'jackfruit':
        print('found jackfruit — stopping the loop')
        break
    print(i)

print('loop finished')
````

````python
# continue — print everything except the number 5

numbers = [3, 4, 6, 2, 54, 6, 3, 5, 7, 5, 3, 57, 5, 34]

for i in numbers:
    if i == 5:
        continue
    print(i, end='  ')

print()
print()
print('in the original list :', len(numbers))
print('how many 5s          :', numbers.count(5))
print('how many printed     :', len(numbers) - numbers.count(5))
````

````python
# another example — skipping banana

fruits = ['apple', 'banana', 'cherry']

for x in fruits:
    if x == 'banana':
        continue
    print(x)
````

### The two side by side

````python
# the same list and condition — only break and continue are swapped

print('using break :')
for n in [1, 2, 3, 4, 5]:
    if n == 3:
        break
    print('   ', n)

print()
print('using continue :')
for n in [1, 2, 3, 4, 5]:
    if n == 3:
        continue
    print('   ', n)
````

# Part 7 — While Loop

A `for` loop runs **a fixed number of times** — as many as there are items.
A `while` loop runs **as long as a condition stays true** — you do not know in advance how many times.

```
i = 1                 ← 1. set the starting value
while i < 5:          ← 2. test the condition
    print(i)          ← 3. do the work
    i += 1            ← 4. change the value  ← forget this and you get an infinite loop!
```

**Never forget the three steps:** starting value → condition → **change the value**.

````python
# a simple while loop

i = 1

while i < 5:
    print(i)
    i += 1          # shorthand for i = i + 1

print('loop finished, i is now', i)
````

### `while … else`

A `while` can have an `else` too — when the condition turns false and the loop
**ends normally**, the `else` block runs once. But if you leave with `break`, the `else` does **not** run.

````python
# when it ends normally, the else runs

i = 1
while i < 5:
    print(i, end='  ')
    i += 1
else:
    print('\ni is no longer less than 5')
````

````python
# leaving with break means the else does not run — see the difference

i = 1
while i < 5:
    if i == 3:
        print('\nleaving with break')
        break
    print(i, end='  ')
    i += 1
else:
    print('this line will not be printed')

print('loop finished')
````

### ⚠ If the condition is false at the start, the loop never runs at all

This is a **very common mistake**. Below, `i` starts at 5, so `i < 5`
is false on the first test — the `print` inside never runs.

````python
# when the starting value is wrong

i = 5

while i < 5:
    print(i)          # <- never runs
    i += 1
else:
    print('the loop body was never entered, because 5 < 5 is false from the start')
````

### ⚠ The infinite loop

Forget the `i += 1` line and `i` never increases, the condition never becomes false —
and the notebook just keeps running. You then have to press the **■ (Interrupt / Stop)** button above.

The cell below deliberately contains that mistake, but a **safety counter**
has been added so that it stops by itself after ten rounds.

````python
# what an infinite loop looks like — demonstrated safely

i = 1
safety = 0

while i < 5:
    print('i is still', i, '— it never increases!')

    safety += 1
    if safety >= 5:
        print('\nThere is no i += 1 line here, so the loop would never stop.')
        print('   In real code you would have to press Interrupt to stop it.')
        break
````

### `while True` — deliberately infinite, then leaving with `break`

This is actually a very useful pattern: keep going until the user says "stop".

````python
# first without input — the same logic, taking values from a list

for n in [3, 4, 2, 7]:
    if n == 2:
        print('got a 2 — stopping')
        break
    print('Square of', n, 'is', n ** 2)
````

### for versus while — which one when?

| | `for` | `while` |
|---|-------|---------|
| How many times | **known in advance** | **not known** |
| Usually driven by | a list or a range | a condition |
| Risk of an infinite loop | almost none | **yes** |
| Example | every row of a dataset | until the user says "stop" |

**The rule:** if there is a list or a definite count → `for`. If not → `while`.

### Splitting with a for loop and if–else

We walk through every value of the `Earth_skin_temp` column and collect them into two
separate lists, according to whether they are above or below 16 °C.

````python
# cold and warm — into two groups

min_1 = []
max_1 = []

skin = list(df.Earth_skin_temp)

for i in skin:
    if i <= 16:
        min_1.append(i)
    else:
        max_1.append(i)

print('total days        :', len(skin))
print('<= 16 C (cold)    :', len(min_1), 'days')
print('>  16 C (warm)    :', len(max_1), 'days')
print()
print('do they add up?   :', len(min_1) + len(max_1) == len(skin))
````

### Splitting into seasons with if–elif–else

We use the `MO` (month) column to assign each day to a season, and a `Dictionary` (Class 03)
to count how many days fall in each.

````python
# month to season

def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Summer'
    elif month in [6, 7, 8, 9]:
        return 'Monsoon'
    else:
        return 'Autumn'


season_count = {}

for m in df.MO:
    s = get_season(m)
    season_count[s] = season_count.get(s, 0) + 1

print('season    days')
print('-' * 18)
for season, days in season_count.items():
    print(f'{season:<8} {days}')

print()
print('total :', sum(season_count.values()), 'days')
````

### The mean temperature of each season

This time we keep **two** dictionaries inside the loop — one for the sum, one for the count.
Dividing at the end gives the mean.

````python
# mean maximum temperature by season

total_temp = {}
day_count = {}

for month, temp in zip(df.MO, df.Max_temp):
    s = get_season(month)
    total_temp[s] = total_temp.get(s, 0) + temp
    day_count[s] = day_count.get(s, 0) + 1

print('season    days     mean maximum temperature')
print('-' * 42)
for s in total_temp:
    avg = total_temp[s] / day_count[s]
    print(f'{s:<8} {day_count[s]:<8} {avg:.2f} °C')
````

# Part 9 — Everything together: building a grade sheet

Now all three of today's ideas at once — a **List** (Class 03) turned into a
**DataFrame**, a **for loop** over it, and an **if–elif–else** inside.

````python
# the students' marks

number_list = [82, 72, 23, 45, 98, 100, 34, 65, 78]
name_list = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I']

import pandas as pd
result = pd.DataFrame({'Number': number_list, 'Name': name_list})

result
````

> **⚠ A mistake people often make:** if you write `result` in a cell and **then** write more code,
> the table is no longer displayed — Jupyter only shows the result of the **last line**.
> To see the table, either put it in its own cell or use `display(result)`.

````python
# working out the grades using a counter called sl (the original approach)

sl = 0

for i in result.Number:
    if i >= 80:
        print(result.Name[sl], '= A+')
    elif i >= 70:
        print(result.Name[sl], '= A')
    elif i >= 60:
        print(result.Name[sl], '= A-')
    elif i >= 50:
        print(result.Name[sl], '= B')
    elif i >= 40:
        print(result.Name[sl], '= C')
    elif i >= 33:
        print(result.Name[sl], '= D')
    else:
        print(result.Name[sl], '= F')

    sl = sl + 1
````

### The same job, written more cleanly

In the code above `sl` has to be incremented by hand — forget it once and every name goes astray.
Taking the names and marks **together** with `zip()` (learned in Class 03) removes that risk entirely.

````python
# zip() plus the get_grade() function made earlier

print('name   mark   grade')
print('-' * 22)

for name, number in zip(result.Name, result.Number):
    print(f'{name:<5} {number:>4}    {get_grade(number)}')
````

````python
# adding the grade to the table as a column

grades = []

for number in result.Number:
    grades.append(get_grade(number))

result['Grade'] = grades
result
````

````python
# how many students got each grade

grade_count = {}

for g in result.Grade:
    grade_count[g] = grade_count.get(g, 0) + 1

for g in ['A+', 'A', 'A-', 'B', 'C', 'D', 'F']:
    if g in grade_count:
        print(f'{g:<3} -> {grade_count[g]} students')

print()
print('passed :', sum(1 for g in result.Grade if g != 'F'), 'students')
print('failed :', sum(1 for g in result.Grade if g == 'F'), 'students')
````

# Today's summary

| What we learned | How it is written | Worth remembering |
|-----------|------------|----------------|
| `bool` | `True` / `False` | capital letter, no quotes. `True` = 1 |
| falsy values | `0`, `''`, `[]`, `{}`, `None` | everything else is truthy |
| `if … else` | colon `:` + four spaces of indent | two paths |
| `if … elif … else` | as many `elif` as you like | **it stops at the first true condition** — order matters |
| Nested if | an `if` inside an `if` | the inner one runs only if the outer is true |
| `for` | `for x in sequence:` | the number of runs is **known** |
| `range()` | `range(start, stop, step)` | starts at 0, **the stop value is excluded** |
| `while` | `while condition:` | the number of runs is **not known** |
| `break` | **stops** the loop | the `else` of `while … else` does not run |
| `continue` | **skips** this step | moves on to the next |
| `%` | remainder | `% 2 == 0` → even |

### The five most common mistakes

1. **Forgetting the colon `:`** → `SyntaxError`
2. **Missing or inconsistent indentation** → `IndentationError`
3. **Putting the `elif` conditions in the wrong order** → grade D for a score of 95
4. **Forgetting to change the value in a `while`** (`i += 1`) → an infinite loop
5. **Not wrapping `input()` in `int()`** → `'56' > '34'` is compared character by character
